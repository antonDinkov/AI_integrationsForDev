```python
from pprint import pprint
from pydantic import BaseModel

def print_response (response):
  print(f"Response id: {response.id}")
  print(f"Output tokens: {response.usage.output_tokens} ({response.usage.output_tokens_details.reasoning_tokens} reasoning tokens); Input tokens: {response.usage.input_tokens} ({response.usage.input_tokens_details.cached_tokens} cached)")
  pprint(response.output)

  print()
  print(f"{'-' * 20} [Text] {'-' * 20}")
  print(response.output_text)
```

```python
from openai import OpenAI
from google.colab import userdata

api_key = userdata.get('OPEN_AI_API_KEY')
client = OpenAI(api_key=api_key)
```

```python
customer:str = "Petar Ivanov"
```

```python
generate_customer_name_response = client.responses.create(
    prompt={
        "id": "pmpt_6a5e1a3aa2648190a9b703d949d1a3550e383bdfd28ef90d",
        "version": "3",
        "variables": {
            "customer": customer
        }
    }
)
```

```python
generate_customer_name_response
```

```python
print_response(generate_customer_name_response)
```
