```python
from openai import OpenAI
from pydantic import BaseModel, Field
from google.colab import userdata

openai_api_key = userdata.get('OPEN_AI_API_KEY')

openai_client = OpenAI(api_key=openai_api_key)
```

```python
response = openai_client.responses.create(
    model="gpt-5.4-nano",
    input="Hi, my name is toni. What is yout name?"
)
```

```python
response
```

```python
conversation_response = openai_client.responses.create(
    model="gpt-5.4-nano",
    instructions="Talk like a pirate",
    previous_response_id =response.id,
    input=[
        {
            "role": "developer",
            "content": "When address me, use 'Mr'"
        },
        {
            "role": "user",
            "content": "Do you remember what was my name?"
        }
    ]
)
```

```python
conversation_response.output_text
```

# Structured output

```python
class CalendarEvent(BaseModel):
  name: str
  short_description: str
  duration_h: float = Field(description="Duration of the event in hours")
```

```python
structured_response = openai_client.responses.parse(
    model="gpt-5.4-nano",
    input="Generate a random calendar event ",
    text_format=CalendarEvent,
)
```

```python
structured_response.output_parsed.duration_h
```

## Streaming

```python
with openai_client.responses.stream(model="gpt-5.4-nano", input="Describe all popular types of pizza with their ingredients and rate them by best flavour.") as basic_stream:
  for event in basic_stream:
    if 'delta' in event.dict():
      print(event.delta, end='', flush=True)
```
