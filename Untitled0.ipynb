```python
from anthropic import Anthropic
from google.colab import userdata
from pydantic import BaseModel, Field

anthropic_api_key = userdata.get('ANTHROPIC_API_KEY')
client = Anthropic(
    api_key=anthropic_api_key
)
```

```python
models = client.models.list()
```

```python
models.data
```

```python
response = client.messages.create(
    model="claude-haiku-4-5-20251001",
    messages=[
        {"role": "user", "content": "Hello"}
    ],
    max_tokens=100
)
```

```python
response
```

```python
introduction_response = client.messages.create(
    model="claude-haiku-4-5-20251001",
    messages=[{"role": "user", "content":"Hi, my name is toni. What is yout name?"}],
    max_tokens=100
)
```

```python
introduction_response.content[0].text
```

```python
acquintance_response = client.messages.create(
    model="claude-haiku-4-5-20251001",
    messages=[
        {"role": "user", "content": "Hi, my name is toni. What is yout name?"},
        {"role": "assistant", "content": introduction_response.content},
        {"role": "user", "content": "Do you remember what was my name?"}
    ],
    max_tokens=1024
)
```

```python
acquintance_response.content[0].text
```

```python
def simplify(response):
  return {"role": "assistant", "content": response.content}

acquintance2_response = client.messages.create(
    model="claude-haiku-4-5-20251001",
    messages=[
        {"role": "user", "content": "Hi, my name is toni. What is yout name?"},
        simplify(introduction_response),
        {"role": "user", "content": "Do you remember what was my name?"}
    ],
    max_tokens=1024,
    system="You speak like a pirate"
)
```

```python
acquintance2_response.content[0].text
```

## Structured Output

```python
class CalendarEvent(BaseModel):
  name: str
  short_description: str
  duration_h: float = Field(description="Duration of the event in hours")
```

```python
structured_response = client.messages.parse(
    model="claude-haiku-4-5-20251001",
    messages=[
        {"role": "user", "content": "Generate a random calendar event"}
    ],
    max_tokens=1024,
    output_format=CalendarEvent
)
```

```python
structured_response.content[0].parsed_output.name
```

## Streaming

```python
with client.messages.stream(
    model="claude-haiku-4-5-20251001",
    messages=[
        {"role": "user", "content": "Describe all popular types of pizza with their ingredients and rate them by best flavour."}
    ],
    max_tokens=4096,
) as basic_stream:
    for event in basic_stream.text_stream:
        print(event, end='', flush=True)
```
